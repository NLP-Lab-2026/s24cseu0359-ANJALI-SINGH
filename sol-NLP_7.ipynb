```python
# ============================================================
# CSET346 - NLP LAB 7
# Sequence Modelling for Text Classification using RNN
# Q1: Text Preprocessing and Sequence Representation
# Q2: Simple RNN Model
# Q3: Simple RNN vs LSTM vs GRU
# Q4: Best Model Evaluation and Error Analysis
# ============================================================

# STEP 0: INSTALL AND IMPORT LIBRARIES
!pip install -q tensorflow scikit-learn matplotlib pandas numpy

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, LSTM, GRU, Dense
from tensorflow.keras.optimizers import Adam

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay
)

np.random.seed(42)
tf.random.set_seed(42)

print("TensorFlow version:", tf.__version__)


# ============================================================
# Q1. TEXT PREPROCESSING AND SEQUENCE REPRESENTATION
# ============================================================

print("\n========== Q1: TEXT PREPROCESSING ==========")

# Dataset configuration
VOCAB_SIZE = 10000
MAX_LEN = 200
EMBEDDING_DIM = 64
RNN_UNITS = 64
BATCH_SIZE = 128
EPOCHS = 3

# Load IMDb dataset
# 0 = Negative sentiment
# 1 = Positive sentiment
(X_train_raw, y_train), (X_test_raw, y_test) = imdb.load_data(
    num_words=VOCAB_SIZE
)

print("Original training reviews:", len(X_train_raw))
print("Original testing reviews:", len(X_test_raw))
print("Vocabulary limit:", VOCAB_SIZE)
print("Maximum sequence length:", MAX_LEN)

# Create reverse word index for decoding reviews
word_index = imdb.get_word_index()

reverse_word_index = {
    index + 3: word
    for word, index in word_index.items()
}

reverse_word_index[0] = "<PAD>"
reverse_word_index[1] = "<START>"
reverse_word_index[2] = "<UNK>"
reverse_word_index[3] = "<UNUSED>"

def decode_review(sequence):
    """Convert an integer sequence back into readable text."""
    return " ".join(
        reverse_word_index.get(int(token), "<UNK>")
        for token in sequence
    )

# Display 5 sample reviews with sentiment labels
print("\n----- Five Sample Reviews -----")

for i in range(5):
    print("\nReview", i + 1)
    print("Text:", decode_review(X_train_raw[i])[:500])
    print("Label:", "Positive" if y_train[i] == 1 else "Negative")

# Padding and truncation
# Pre-padding and pre-truncation keep the most recent tokens
X_train_padded = pad_sequences(
    X_train_raw,
    maxlen=MAX_LEN,
    padding="post",
    truncating="post"
)

X_test_padded = pad_sequences(
    X_test_raw,
    maxlen=MAX_LEN,
    padding="post",
    truncating="post"
)

# Split training data into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(
    X_train_padded,
    y_train,
    test_size=0.2,
    random_state=42,
    stratify=y_train
)

print("\n----- Dataset Shapes -----")
print("Training data:", X_train.shape)
print("Training labels:", y_train.shape)
print("Validation data:", X_val.shape)
print("Validation labels:", y_val.shape)
print("Testing data:", X_test_padded.shape)
print("Testing labels:", y_test.shape)

# Show original text, integer sequence and padded sequence
print("\n----- Original and Numerical Sequences -----")

for i in range(3):
    print(f"\nReview {i + 1}")
    print("Original review:", decode_review(X_train_raw[i])[:300])
    print("Integer sequence:", X_train_raw[i][:30])
    print("Padded sequence:", X_train_padded[i][:30])
    print("Padded length:", len(X_train_padded[i]))

print("\nQ1 Observations:")
print("IMDb is a binary sentiment classification dataset.")
print("Words are represented by integer IDs.")
print("Padding makes every sequence the same length.")
print("Label 0 = Negative; Label 1 = Positive.")


# ============================================================
# HELPER FUNCTION: BUILD RNN MODELS
# ============================================================

def build_model(model_type):
    """
    Create a sentiment classification model:
    Embedding -> RNN/LSTM/GRU -> Dense sigmoid output.
    """
    model = Sequential(name=model_type + "_Sentiment_Model")

    model.add(
         Embedding(
            input_dim=VOCAB_SIZE,
            output_dim=EMBEDDING_DIM,
            mask_zero=True,
            name="Embedding"
        )
    )

    if model_type == "SimpleRNN":
        model.add(
            SimpleRNN(RNN_UNITS, name="Simple_RNN")
        )

    elif model_type == "LSTM":
        model.add(
            LSTM(RNN_UNITS, name="LSTM")
        )

    elif model_type == "GRU":
        model.add(
            GRU(RNN_UNITS, name="GRU")
        )

    model.add(Dense(1, activation="sigmoid", name="Output"))

    model.compile(
        optimizer=Adam(learning_rate=0.001),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    return model


# ============================================================
# Q2. SIMPLE RNN SENTIMENT CLASSIFICATION
# ============================================================

print("\n========== Q2: SIMPLE RNN MODEL ==========")

simple_rnn_model = build_model("SimpleRNN")

print("\n----- Simple RNN Architecture -----")
simple_rnn_model.summary()

# Train Simple RNN
history_rnn = simple_rnn_model.fit(
    X_train,
    y_train,
    validation_data=(X_val, y_val),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    verbose=1
)

# Plot training and validation accuracy
plt.figure(figsize=(8, 5))
plt.plot(history_rnn.history["accuracy"], label="Training Accuracy")
plt.plot(history_rnn.history["val_accuracy"], label="Validation Accuracy")
plt.title("Simple RNN: Training vs Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.show()

# Plot training and validation loss
plt.figure(figsize=(8, 5))
plt.plot(history_rnn.history["loss"], label="Training Loss")
plt.plot(history_rnn.history["val_loss"], label="Validation Loss")
plt.title("Simple RNN: Training vs Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.show()

# Evaluate Simple RNN on test data
rnn_test_loss, rnn_test_accuracy = simple_rnn_model.evaluate(
    X_test_padded,
    y_test,
    verbose=0
)

print("\nSimple RNN Test Loss:", round(rnn_test_loss, 4))
print("Simple RNN Test Accuracy:", round(rnn_test_accuracy, 4))


# ============================================================
# Q3. COMPARE SIMPLE RNN, LSTM AND GRU
# ============================================================

print("\n========== Q3: MODEL COMPARISON ==========")

models = {
    "SimpleRNN": simple_rnn_model
}

histories = {
    "SimpleRNN": history_rnn
}

# Train LSTM and GRU using exactly the same data split
for model_name in ["LSTM", "GRU"]:
    print(f"\n----- Training {model_name} -----")

    model = build_model(model_name)
    model.summary()

    history = model.fit(
        X_train,
        y_train,
        validation_data=(X_val, y_val),
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        verbose=1
    )

    models[model_name] = model
    histories[model_name] = history

# Evaluate all models using the same test set
results = []

for model_name, model in models.items():
    print(f"\nEvaluating {model_name}...")

    probabilities = model.predict(
        X_test_padded,
        batch_size=BATCH_SIZE,
        verbose=0
    ).ravel()

    predictions = (probabilities >= 0.5).astype(int)

    accuracy = accuracy_score(y_test, predictions)
    precision = precision_score(
        y_test, predictions, zero_division=0
    )
    recall = recall_score(
        y_test, predictions, zero_division=0
    )
    f1 = f1_score(
        y_test, predictions, zero_division=0
    )

    results.append({
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1
    })

    print(classification_report(
        y_test,
        predictions,
        target_names=["Negative", "Positive"],
        zero_division=0
    ))

comparison_df = pd.DataFrame(results)
comparison_df = comparison_df.sort_values(
    by="F1-Score",
    ascending=False
).reset_index(drop=True)

print("\n----- Model Performance Comparison -----")
print(comparison_df.round(4).to_string(index=False))

# Visual comparison of test metrics
comparison_df.set_index("Model")[
    ["Accuracy", "Precision", "Recall", "F1-Score"]
].plot(kind="bar", figsize=(10, 6))

plt.title("Simple RNN vs LSTM vs GRU")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.grid(axis="y")
plt.tight_layout()
plt.show()

# Plot training and validation accuracy for each architecture
plt.figure(figsize=(9, 6))

for model_name, history in histories.items():
    plt.plot(
        history.history["val_accuracy"],
        marker="o",
        label=model_name + " Validation"
    )

plt.title("Validation Accuracy Comparison")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.show()

# Generate predictions for at least 10 test reviews
print("\n----- Predictions for 10 Test Reviews -----")

best_model_name = comparison_df.iloc[0]["Model"]
best_model = models[best_model_name]

test_probabilities = best_model.predict(
    X_test_padded,
    batch_size=BATCH_SIZE,
    verbose=0
).ravel()

test_predictions = (test_probabilities >= 0.5).astype(int)

prediction_rows = []

for i in range(10):
    actual = int(y_test[i])
    predicted = int(test_predictions[i])

    prediction_rows.append({
        "Review": decode_review(X_test_raw[i])[:250],
        "Actual": "Positive" if actual == 1 else "Negative",
        "Predicted": "Positive" if predicted == 1 else "Negative",
        "Confidence": round(
            float(
                test_probabilities[i]
                if predicted == 1
                else 1 - test_probabilities[i]
            ),
            4
        )
    })

print(pd.DataFrame(prediction_rows).to_string(index=False))


# ============================================================
# Q4. BEST MODEL EVALUATION AND ERROR ANALYSIS
# ============================================================

print("\n========== Q4: ERROR ANALYSIS ==========")

# Select the model with the highest test F1-score
# For a stricter experimental protocol, select using validation
# metrics and use the test set only for final evaluation.
best_model_name = comparison_df.iloc[0]["Model"]
best_model = models[best_model_name]

print("Selected best model:", best_model_name)

# Final predictions
best_probabilities = best_model.predict(
    X_test_padded,
    batch_size=BATCH_SIZE,
    verbose=0
).ravel()

best_predictions = (best_probabilities >= 0.5).astype(int)

# Final evaluation metrics
accuracy = accuracy_score(y_test, best_predictions)
precision = precision_score(
    y_test, best_predictions, zero_division=0
)
recall = recall_score(
    y_test, best_predictions, zero_division=0
)
f1 = f1_score(
    y_test, best_predictions, zero_division=0
)

print("\n----- Final Evaluation Metrics -----")
print("Accuracy :", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall   :", round(recall, 4))
print("F1-score :", round(f1, 4))

print("\nDetailed Classification Report:")
print(classification_report(
    y_test,
    best_predictions,
    target_names=["Negative", "Positive"],
    zero_division=0
))

# Confusion matrix
cm = confusion_matrix(
    y_test,
    best_predictions,
    labels=[0, 1]
)

print("\nConfusion Matrix:")
print(cm)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Negative", "Positive"]
)

disp.plot(cmap="Blues", values_format="d")
plt.title(f"Confusion Matrix - {best_model_name}")
plt.show()

# Identify misclassified reviews
wrong_indices = np.where(best_predictions != y_test)[0]

print("\nNumber of misclassified reviews:", len(wrong_indices))
print("\n----- Five Incorrectly Classified Reviews -----")

# Possible linguistic clues to help manual error analysis.
# These are heuristic clues, not proof of the model's reasoning.
error_clues = {
    "Negation": [
        "not", "never", "no", "isn't", "wasn't",
        "didn't", "hardly", "without"
    ],
    "Sarcasm or irony": [
        "yeah right", "what a joke", "so-called",
        "wonderful", "brilliant", "just great"
    ],
    "Mixed sentiment": [
        "but", "however", "although", "despite",
        "on the other hand"
    ],
    "Intensifiers or ambiguity": [
        "very", "really", "quite", "somewhat",
        "surprisingly", "oddly"
    ]
}

def suggest_error_clues(review):
    review_lower = review.lower()
    matched = []

    for category, keywords in error_clues.items():
        if any(keyword in review_lower for keyword in keywords):
            matched.append(category)

    if not matched:
        matched.append(
            "Possible context, vocabulary or long-distance dependency issue"
        )

    return ", ".join(matched)

for count, idx in enumerate(wrong_indices[:5], start=1):
    review = decode_review(X_test_raw[idx])
    actual = int(y_test[idx])
    predicted = int(best_predictions[idx])
    probability = float(best_probabilities[idx])

    print(f"\nIncorrect Review {count}")
    print("Review:", review[:1000])
    print("Actual:", "Positive" if actual == 1 else "Negative")
    print("Predicted:", "Positive" if predicted == 1 else "Negative")
    print("Positive probability:", round(probability, 4))
    print("Possible error clues:", suggest_error_clues(review))

# Save all test predictions for further analysis
all_results = pd.DataFrame({
    "Review": [decode_review(seq) for seq in X_test_raw],
    "Actual": y_test,
    "Predicted": best_predictions,
    "Positive_Probability": best_probabilities,
    "Correct": best_predictions == y_test
})

all_results.to_csv(
    "imdb_rnn_predictions.csv",
    index=False,
    encoding="utf-8-sig"
)

comparison_df.to_csv(
    "imdb_model_comparison.csv",
    index=False
)

print("\nSaved files:")
print("1. imdb_rnn_predictions.csv")
print("2. imdb_model_comparison.csv")

print("\n========== LAB 7 COMPLETED ==========")
```
